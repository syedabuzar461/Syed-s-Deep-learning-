```
import torch
import torch.nn as nn
from torchvision import models

class DensePixelNet(nn.Module):
    def __init__(self, num_classes=5):
        super(DensePixelNet, self).__init__()

        # DenseNet121 feature extractor
        densenet = models.densenet121(
            weights=models.DenseNet121_Weights.DEFAULT
        )

        self.features = densenet.features

        # Pixel-level segmentation decoder
        self.decoder = nn.Sequential(
            nn.Conv2d(1024, 512, 3, padding=1),
            nn.ReLU(),

            nn.ConvTranspose2d(
                512, 256, 2, stride=2
            ),
            nn.ReLU(),

            nn.ConvTranspose2d(
                256, 128, 2, stride=2
            ),
            nn.ReLU(),

            nn.ConvTranspose2d(
                128, 64, 2, stride=2
            ),
            nn.ReLU(),

            nn.ConvTranspose2d(
                64, num_classes, 2, stride=2
            )
        )

    def forward(self, x):
        x = self.features(x)
        x = nn.functional.relu(x)
        x = self.decoder(x)
        return x


# Five road-scene classes
model = DensePixelNet(num_classes=5)

# Test image
image = torch.randn(1, 3, 224, 224)

output = model(image)

print("DenseNet + PixelNet Architecture")
print("---------------------------------")
print("Input shape :", image.shape)
print("Output shape:", output.shape)

print("\nPixel Classes:")
print("0 - Road")
print("1 - Pedestrian")
print("2 - Vehicle")
print("3 - Traffic Sign")
print("4 - Background")
```
