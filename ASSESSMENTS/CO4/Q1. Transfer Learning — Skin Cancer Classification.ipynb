```
import torch
import torch.nn as nn
from torchvision import models

# Load pretrained ResNet50
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)

# Freeze all layers
for param in model.parameters():
    param.requires_grad = False

# Replace final classification layer
num_features = model.fc.in_features

model.fc = nn.Sequential(
    nn.Linear(num_features, 256),
    nn.ReLU(),
    nn.Dropout(0.5),
    nn.Linear(256, 2)
)

# Loss function
criterion = nn.CrossEntropyLoss()

# Optimizer only for new classification layers
optimizer = torch.optim.Adam(
    model.fc.parameters(),
    lr=0.001
)

print("Transfer Learning Model")
print("-----------------------")
print(model.fc)

print("\nFrozen layers: Early and middle ResNet layers")
print("Fine-tuned layers: Final fully connected layers")
print("Number of classes: 2")
print("Classes: Benign, Malignant")
```
